# STEP 4: Perform RFM Calculations

```python
referenceDate = consolidated_data['TransactionDate'].max() + pd.Timedelta(days=1)
referenceDate
```

```python
consolidated_data.groupby(['CustomerID'])['TransactionDate'].max()
```

```python
consolidated_data.groupby(['CustomerID'])['TransactionDate'].count()
```

```python
consolidated_data.groupby(['CustomerID'])['TransactionAmount'].sum()
```

```python
df_rfm = pd.DataFrame()
```

```python
df_rfm = consolidated_data.groupby(['CustomerID']).agg(LastTransactionDay=('TransactionDate','max'),CountOfTransactions=('TransactionDate','count'),SumOfTransactionAmount=('TransactionAmount','sum')).reset_index()
df_rfm
```

### Use a reference date to compute Recency in number of days

```python
df_rfm['Recency'] = referenceDate-df_rfm['LastTransactionDay']
df_rfm
```

### Store result in a new DataFrame called df_rfm

```python
df_rfm = df_rfm.rename(columns={'LastTransactionDay':'Recency',
                                'CountOfTransactions':'Frequency',
                                'SumOfTransactionAmount':'Monetary',
                                'Recency':'RecencyInDays'})
df_rfm
```

# STEP 5: Score RFM

```python
df_rfm['R_Score'] = pd.qcut(df_rfm['RecencyInDays'],5,labels=[5,4,3,2,1])
df_rfm['F_Score'] = pd.qcut(df_rfm['Frequency'].rank(),5,labels=[1,2,3,4,5])
df_rfm['M_Score'] = pd.qcut(df_rfm['Monetary'],5,labels=[1,2,3,4,5])
df_rfm
```

# STEP 6: Create Combined RFM Segment

```python
df_rfm['Segment'] = df_rfm['R_Score'].astype('string')+df_rfm['F_Score'].astype('string')+df_rfm['M_Score'].astype('string')
df_rfm
```

```python
df_rfm.dtypes
```

# Step 7: Assign Segment Labels

```python
conditions = [
    df_rfm['Segment'].str.match(r'^[45][45][45]'),#Champions
    df_rfm['Segment'].str.match(r'^[2345][45].'),#Loyal Customer
    df_rfm['Segment'].str.match(r'^[12][345].'),#At Risk
    df_rfm['Segment'].str.match(r'^[34][23][45]'),#Hibernating
    df_rfm['Segment'].str.match(r'^[45][23].'),#Potential loyalist
    df_rfm['Segment'].str.match(r'^[1][12][12]')]#Lost
labels = [
    'Champions',
    'Loyal Customer',
    'At Risk',
    'Hibernating',
    'Potential loyalist',
    'Lost',]
df_rfm['SegmentLabel'] = np.select(conditions,labels,default = 'Visitor')
df_rfm
```

```python
print(df_rfm[df_rfm['SegmentLabel'] == 'Visitor']['Segment'].unique())
```

# STEP 8: Visualise

### Count of customers in each segment

```python
countOfCustomerForEachSegment = df_rfm.groupby(['SegmentLabel'])[['CustomerID']].count().reset_index()
countOfCustomerForEachSegment
```

```python
plt.figure(figsize = (10,4))
sns.barplot(x = countOfCustomerForEachSegment['SegmentLabel'], y = countOfCustomerForEachSegment['CustomerID'],color = 'cyan',edgecolor = '#005b66')
plt.title('Customer Distribution Across Segments', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('SEGMENTS', fontsize=12, fontweight='bold', labelpad=10)
plt.ylabel("Count Of Customer Id's", fontsize=12, fontweight='bold', labelpad=10)
plt.tight_layout()
plt.show()
```

### Revenue contribution per segment

```python
RevenueContriPerSeg = df_rfm.groupby(['SegmentLabel'])[['Monetary']].sum()
RevenueContriPerSeg = RevenueContriPerSeg.sort_values(by='Monetary', ascending=False)
RevenueContriPerSeg
```

```python
plt.figure(figsize=(10,6))
colors = sns.color_palette('Purples_r', len(RevenueContriPerSeg))
plt.pie(RevenueContriPerSeg['Monetary'], labels = RevenueContriPerSeg.index, autopct = '%1.1f%%',colors = colors,startangle= 140,pctdistance=0.80,textprops={'fontsize': 11, 'weight': 'bold'}, wedgeprops={'edgecolor': 'black', 'linewidth': 1})
plt.title('Revenue Contribution by Segment', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```

### Recency vs Monetary scatter plot colored by segment

```python
plt.figure(figsize =(10,6))
ax = sns.scatterplot(x = df_rfm['RecencyInDays'], y = df_rfm['Monetary'],hue = df_rfm['SegmentLabel'])
ax.yaxis.set_major_formatter(ticker.StrMethodFormatter('₹{x:,.0f}'))
```

```python
fig, ax = plt.subplots(1,2,figsize=(14,8))
colors1 = sns.color_palette('viridis', len(countOfCustomerForEachSegment))
colors2 = sns.color_palette('Purples_r', len(countOfCustomerForEachSegment))
ax[0].pie(countOfCustomerForEachSegment['CustomerID'], labels = countOfCustomerForEachSegment['SegmentLabel'],colors = colors1, autopct = '%1.1f%%',startangle= 140,pctdistance=0.82,textprops={'fontsize': 11, 'weight': 'bold'}, wedgeprops={'edgecolor': 'white', 'linewidth': 1})
ax[1].pie(RevenueContriPerSeg['Monetary'], labels = RevenueContriPerSeg.index, autopct = '%1.1f%%',colors = colors2,startangle= 140,pctdistance=0.82,textprops={'fontsize': 11, 'weight': 'bold'}, wedgeprops={'edgecolor': 'white', 'linewidth': 1})
ax[0].set_title('Percentage of Customers in each Segment', fontsize = 14, fontweight = 'bold', pad =20)
ax[1].set_title('Revenue Contribution by Segment', fontsize = 14, fontweight = 'bold', pad = 20)
plt.tight_layout()
plt.show()
```

```python
if 'SegmentLabel' in countOfCustomerForEachSegment.columns:
    count_df = countOfCustomerForEachSegment.set_index('SegmentLabel')
else:
    count_df = countOfCustomerForEachSegment.copy()

df_analysis = pd.DataFrame({
    'Revenue': RevenueContriPerSeg['Monetary'],
    'CountOfCustomer': count_df['CustomerID']
}).sort_values(by='Revenue', ascending=False)

df_analysis['Cum_Revenue'] = df_analysis['Revenue'].cumsum()
df_analysis['Cum_Customers'] = df_analysis['CountOfCustomer'].cumsum()

df_analysis['Cum_Revenue_%'] = (df_analysis['Cum_Revenue'] / df_analysis['Revenue'].sum()) * 100
df_analysis['Cum_Customers_%'] = (df_analysis['Cum_Customers'] / df_analysis['CountOfCustomer'].sum()) * 100

styled_df = df_analysis.style.format({
    'Revenue': '₹{:,.2f}',
    'CountOfCustomer': '{:,.0f}',
    'Cum_Revenue': '₹{:,.2f}',
    'Cum_Customers': '{:,.0f}',
    'Cum_Revenue_%': '{:.1f}%',
    'Cum_Customers_%': '{:.1f}%'
})

styled_df
```

1. Since highest revenue comes from Visitor section where either Recency, Frequency or Monetary has the lowest score, We can make offers available for first time consumers to increace there RFM score.
2. To retain Champions and Loyalists exclusive offers can be provided to retain there trust
3. Visitors having higher monetary scores but lower recency or frequency. Thus they can be sent messages informing them of various offers
4. Hibernationg customers can be sent messages to revive there account
5. Loyal customers and Champions follow Visitors in revenue. Thus it becomes important to inform them of any upcoming sales beforehand.
6. Scatterplot displays that substantial amount of transaction happens where Recency has the lowest score. Thus most orders under every segments gets placed after a gap of few days
7. Loyal customers and Champions spends the highests among all the others
8. Visitors spend the lowest in aggregate, among all the groups
9. Loyal customers expenses are far more equally distributed when it comes to expenses
10. 73% of the customer generate 76% of the revenue. Hence revenue contribution is uniform leaving space for different segments to upgrade to Loyalist or Champions
